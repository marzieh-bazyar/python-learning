```python
import random
com_num= random.randint(1,20)
count_guess = 0
guesses = []
number = int(input("🎯Enter your guess between 1 to 20  :"))
if number in guesses:
    print("Oops! You have already guessed this number.")
else:
    guesses.append(number)
    count_guess+=1
while number != com_num  and count_guess < 5:
     if number < com_num:
        print("😣Too low")
     elif number > com_num:
        print("😁Too high")
     number = int(input("🎯Enter your guess between 1 to 20  :"))
     count_guess+=1
     if number in guesses:
          print("Oops! You have already guessed this number!")
     else:
           guesses.append(number)
           count_guess+=1
print("-"*45)
if number==com_num:
   print('🎉Congrats!!You win')
else:
   print("😔Sorry! you have no chances to guess:  ")
print(f'The correct number was{com_num}')
print('*' * 45)
print(f'Your guessed numbers are : ' , guesses)
print(f'Number of guesses: {count_guess}')

```
