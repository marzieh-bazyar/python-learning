```python
import random
board = ["1", "2", "3",
         "4", "5", "6",
         "7", "8", "9"]
current_player = random.choice(["X", "O"])
game_active = True
moves_made = 0

print("Welcome to Tic-Tac-Toe!")

player_1 = input("Player 1👩‍🎓 enter your name: ")
player_2 = input("Player 2👨‍🎓 enter your name: ")

if current_player == "X":
    print(f"{player_1} starts as X")
    player_x = player_1
    player_o = player_2
else:
    print(f"{player_2} starts as O")
    player_x = player_2
    player_o = player_1

print("Guide: Enter a number from 1 to 9 to make your move.\n")


win_conditions = [
    (0, 1, 2), (3, 4, 5), (6, 7, 8),
    (0, 3, 6), (1, 4, 7), (2, 5, 8),
    (0, 4, 8), (2, 4, 6)
]


while game_active:

    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("-----------")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("-----------")
    print(f" {board[6]} | {board[7]} | {board[8]} \n")


    if current_player == "X":
        current_name = player_1
    else:
        current_name = player_2


    user_input = input(
        f"{current_name}'s turn. Choose a cell (1-9): "
    )


    if not user_input.isdigit():
        print("Invalid input!\n")
        continue


    choice = int(user_input) - 1


    if choice < 0 or choice > 8:
        print("Out of range!\n")
        continue


    if board[choice] == "X" or board[choice] == "O":
        print("This cell is already taken!\n")
        continue


    board[choice] = current_player
    moves_made += 1


    player_won = False

    for cond in win_conditions:
        if (board[cond[0]] == board[cond[1]] ==
            board[cond[2]] == current_player):

            player_won = True
            break


    if player_won:

        print(f"\n {board[0]} | {board[1]} | {board[2]} ")
        print("-----------")
        print(f" {board[3]} | {board[4]} | {board[5]} ")
        print("-----------")
        print(f" {board[6]} | {board[7]} | {board[8]} \n")


        if current_player == "X":
            winner = player_1
        else:
            winner = player_2


        print(f"🎉 Congratulations {winner}! You win!")
        game_active = False


    elif moves_made == 9:

        print("It's a draw!")
        game_active = False


    else:

        answer = input(
            "Would you like to continue? (yes/no): "
        )


        if answer.lower() in ["no", "n"]:

            game_active = False
            print("Game over. Thanks for playing!")


        else:

            if current_player == "X":
                current_player = "O"
            else:
                current_player = "X"
```
