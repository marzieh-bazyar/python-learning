```python
def calculate_average(numbers):
    """
    Calculate and return the average of a list of numbers.
    If the list is empty, return 0.
    """
    if not numbers:
        return 0
    else:        return sum(numbers) / len(numbers)

numbers = list(map(float, input("Please enter numbers separated by spaces: ").split()))

print(f"{calculate_average(numbers) : .2f} ")
```
