```python
def count_vowels(text):
    """
    Count the number of vowels in a string.
    Return the total count.
    """
    vowels = "aeiouAEIOU"
    count = 0

    for letter in text:
        if letter in vowels:
            count += 1

    return count

print(count_vowels("Hello Python, HELLO"))
```
