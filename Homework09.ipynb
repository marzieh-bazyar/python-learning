```python
def find_max(numbers):

    if not numbers:
        return None

    max_number = numbers[0]

    for number in numbers:

        if number > max_number:

            max_number = number

    return max_number


numbers = [547, 369, 58, 520, 9, 456, 369, 56]

print(find_max(numbers))
```
