```python
board = ["1", "2", "3",
         "4", "5", "6",
         "7", "8", "9"]

current_player = "X"
game_active = True
moves_made = 0
name = {}
char_game = ["X", "O"]
print("Welcome to Tic-Tac-Toe! ")
for i in range(len(char_game)):
    name_str = input(f"Enter name of the player {i+1} : ")
    name[char_game[i]] = name_str
print("Guide: Enter a number from 1 to 9 to make your move.\n")

# حلقه اصلی بازی
while game_active:
    do_continue = input("Would you like to continue? (Yes/No): ").strip().lower()

    if do_continue in ["no", "n"]:
        print("End game...")
        break

    elif do_continue in ["yes", "y"]:
        print("Continue the game!")
    else:
        print("Invalid input! Please enter Yes/No or Y/N.")
        continue

    # نمایش صفحه بازی
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("-----------")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("-----------")
    print(f" {board[6]} | {board[7]} | {board[8]} \n")

    # گرفتن ورودی از کاربر
    user_input = input(f"Player ({current_player}) turn. Choose a cell (1-9): ")

    # بررسی عددی بودن ورودی بدون استفاده از try-except (با شرط ساده)
    if  not user_input.isdigit():
        print("Invalid input! Please enter a number between 1 and 9.\n")
        continue

    choice = int(user_input) - 1

    # بررسی محدوده عدد
    if choice < 0 or choice > 8:
        print("Out of range! Please choose a number from 1 to 9.\n")
        continue

    # بررسی خالی بودن خانه انتخاب شده
    if board[choice] == "X" or board[choice] == "O":
        print("This cell is already taken! Try another one.\n")
        continue

    # ثبت حرکت

    board[choice] = current_player
    moves_made += 1

    # بررسی شرط‌های برنده شدن
    win_conditions = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),  # افقی
        (0, 3, 6), (1, 4, 7), (2, 5, 8),  # عمودی
        (0, 4, 8), (2, 4, 6)              # قطری
    ]

    player_won = False
    for cond in win_conditions:
        if board[cond[0]] == board[cond[1]] == board[cond[2]] == current_player:
            player_won = True
            break

    if player_won:
        # نمایش صفحه نهایی
        print(f"\n {board[0]} | {board[1]} | {board[2]} ")
        print("-----------")
        print(f" {board[3]} | {board[4]} | {board[5]} ")
        print("-----------")
        print(f" {board[6]} | {board[7]} | {board[8]} \n")
        print(f" Congratulations! Player ({name[current_player]}) wins! ")
        game_active = False
    elif moves_made == 9:
        print("It's a draw! Game over. ")
        game_active = False
    else:
        # تغییر نوبت
        if current_player == "X":
            current_player = "O"
        else:
            current_player = "X"

```
