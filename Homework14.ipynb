```python
numbers =[12, 16, 18, 25, 105, 163, 50, 36, 72]
even_number = []
def count_even(even_number):
    for number in numbers:
         if number % 2 == 0:
             even_number.append(number)
    return even_number
print(len(count_even(even_number)))
```
