```python
numbers = [12, 16, 18, 25, 105, 163, 50, 36, 72]

def is_even(number):
    if number % 2 == 0:
        return True
    else:
        return False


def count_even(numbers):
    even_number = []

    for number in numbers:
        if is_even(number):
            even_number.append(number)

    return len(even_number)


print(count_even(numbers))
```
