```python
x = 0
while x <= 5:
    if x == 3:
        print(x)
    x += 1

for x in range(6):
    if x == 3: print(x)
numbers = [12, 5, 8, 12, 7, 5, 20, 8]
even = 0
odd = 0
for num in numbers:
    if num % 2 == 0:
        even += num
        print(f'{num} is even')
    else:
        odd += num
        print(f'{num} is odd')

print(f'Sum of evens: {even}')
print(f'Sum of odds: {odd}')
numbers = [12, 5, 8, 12, 7, 5, 20, 8]
even = []
odd = []
for num in numbers:
    if num % 2 == 0:
        even.append(num)
        #print(f'{num} is even')
    else:
        odd.append(num)
        #print(f'{num} is odd')

if sum(even) > sum(odd):
    print('even is big')
elif sum(even) < sum(odd):
    print('odd is big')
else: print('odd equal even')

print(f'Sum of evens: {sum(even)}')
print(f'Sum of odds: {sum(odd)}')

numbers = [1, 2, 3, 2, 4, 5, 1, 6]

duplicate = []
unique = []
for num in numbers:
    if numbers.count(num) > 1:
        duplicate.append(num)
        #print(f'{num}: duplicate')
    else:
        unique.append(num)
        #print(f'{num}: unique')

print(set(duplicate))
print(set(unique))
text = input("Enter a text: ")
letters = 0
digits = 0
others = 0
sp = 0
for ch in text:
    if ch.isalpha():
        letters += 1
    elif ch.isdigit():
        digits += 1
    elif ch == '#' or ch == '!' or ch == '&':
        sp += 1
    else:
        others += 1

print("Letters:", letters)
print("Digits:", digits)
print("Special:", sp)
print("Other characters:", others)
```
